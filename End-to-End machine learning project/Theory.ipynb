```python
import json, os
from ipywidgets import Button, HBox, VBox, Label, Layout, HTML
from IPython.display import display

SAVE_FILE = "checklist_status.json"

items = [
    "Frame the Problem and look at the big picture",
    "Get the Data",
    "Explore the Data",
    "Prepare the Data",
    "Shortlist promising models",
    "Fine Tune the system",
    "Present your solution",
    "Launch!"
]

# cycle order: click moves to the next state
cycle = ["Not started", "In progress", "Completed", "To be noted"]
colors = {
    "Not started": "#32cd32",
    "In progress": "#ffeb3b",
    "Completed":   "#e53935",
    "To be noted": "#ffffff",
}
text_colors = {
    "Not started": "black",
    "In progress": "black",
    "Completed":   "white",
    "To be noted": "black",
}

if os.path.exists(SAVE_FILE):
    with open(SAVE_FILE) as f:
        saved_state = json.load(f)
else:
    saved_state = {item: "Not started" for item in items}

buttons = {}
progress_label = Label()
progress_bar = HTML()

def style_button(btn, status):
    btn.description = status
    btn.style.button_color = colors[status]
    btn.style.text_color = text_colors[status]

def update_progress():
    done = sum(1 for item in items if buttons[item].description == "Completed")
    pct = int((done / len(items)) * 100)
    progress_label.value = f"Progress: {done} / {len(items)} completed"
    progress_bar.value = f"""
    <div style="height:10px; width:300px; background:#eee; border-radius:999px; overflow:hidden;">
        <div style="height:100%; width:{pct}%; background:#e53935; border-radius:999px;"></div>
    </div>
    """

def save_state():
    state = {item: buttons[item].description for item in items}
    with open(SAVE_FILE, "w") as f:
        json.dump(state, f, indent=2)

def on_click(btn, item):
    current = btn.description
    next_status = cycle[(cycle.index(current) + 1) % len(cycle)]
    style_button(btn, next_status)
    save_state()
    update_progress()

rows = []
for item in items:
    btn = Button(layout=Layout(width="160px", height="30px"))
    style_button(btn, saved_state.get(item, "Not started"))
    btn.on_click(lambda b, item=item: on_click(buttons[item], item))
    buttons[item] = btn
    label = Label(item, layout=Layout(width="360px"))
    rows.append(HBox([label, btn]))

update_progress()

title = HTML("<h3>ML Project Checklist</h3>")
display(VBox([title] + rows + [progress_label, progress_bar]))
```

<script>
const items = [
  "Frame the Problem and look at the big picture",
  "Get the Data",
  "Explore the Data",
  "Prepare the Data",
  "Shortlist promising models",
  "Fine Tune the system",
  "Present your solution",
  "Launch!"
];

const cycle = ["notstarted", "progress", "done", "note"];
const labels = { notstarted: "Not started", progress: "In progress", done: "Completed", note: "To be noted" };
const colors = { notstarted: ["#32cd32","#000"], progress: ["#ffeb3b","#000"], done: ["#e53935","#fff"], note: ["#ffffff","#000"] };

const STORAGE_KEY = "ml_checklist_status";
let status = JSON.parse(localStorage.getItem(STORAGE_KEY) || "{}");

function render() {
  const ul = document.getElementById("checklist");
  ul.innerHTML = "";
  items.forEach((text, i) => {
    const s = status[i] || "notstarted";
    const [bg, fg] = colors[s];
    const li = document.createElement("li");
    li.innerHTML = `<span class="item-text">${text}</span>
      <button class="status-btn" style="background:${bg}; color:${fg};" data-i="${i}">${labels[s]}</button>`;
    ul.appendChild(li);
  });
  document.querySelectorAll(".status-btn").forEach(btn => {
    btn.addEventListener("click", () => cycleStatus(btn.dataset.i));
  });
  updateProgress();
}

function cycleStatus(i) {
  const current = status[i] || "notstarted";
  const next = cycle[(cycle.indexOf(current) + 1) % cycle.length];
  status[i] = next;
  localStorage.setItem(STORAGE_KEY, JSON.stringify(status));
  render();
}

function updateProgress() {
  const done = items.filter((_, i) => status[i] === "done").length;
  document.getElementById("progress-text").textContent = `${done} / ${items.length} completed`;
  document.getElementById("bar-fill").style.width = `${(done / items.length) * 100}%`;
}

render();
