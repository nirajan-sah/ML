```python
#Step 1: Importing modules
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor

#Downloading and preparing the data
data_root = "https://github.com/ageron/data/raw/main/"
lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")
X = lifesat[["GDP per capita (USD)"]].values
y = lifesat[["Life satisfaction"]].values

#Visualize the data
lifesat.plot(kind='scatter', grid=True, x = "GDP per capita (USD)", y = "Life satisfaction")
plt.axis([23_500, 62_500, 4, 9.5])
plt.show()

#Select a linear model
model = LinearRegression()
model_K = KNeighborsRegressor(n_neighbors = 3)

#Train the model
model.fit(X, y)
model_K.fit(X, y)

#make a prediction for cyprus
X_new = [[37_655.2]]
print("Value printed using model", model.__class__.__name__, model.predict(X_new))
print("Value printed using model", model_K.__class__.__name__, model_K.predict(X_new))
```
